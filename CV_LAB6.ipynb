```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
def show(img,slicing_wo_bg,slicing_w_bg):
  plt.figure(figsize=(8,6))
  plt.subplot(2,2,1)
  plt.imshow(img,cmap='gray')
  plt.title('Original Grayscale image')
  plt.axis('off')
  plt.subplot(2,2,2)
  plt.imshow(slicing_wo_bg,cmap='gray')
  plt.title('slicing without background')
  plt.axis('off')
  plt.subplot(2,2,3)
  plt.imshow(slicing_w_bg,cmap='gray')
  plt.title('slicing with background')
  plt.axis('off')
  plt.tight_layout()
  plt.show()
img=cv2.imread('/content/img.jpg',0)
if img is None:
  print('Error image not found')
else:
  rmin=100
  rmax=200
  slicing_wo_bg=np.zeros(img.shape)
  slicing_w_bg=img.copy()
  for i in range(img.shape[0]):
    for j in range(img.shape[1]):
      pixel_value=img[i][j]
      if rmin<=pixel_value<=rmax:
        slicing_wo_bg[i][j]=255
        slicing_w_bg[i][j]=255
  show(img,slicing_wo_bg,slicing_w_bg)
```
