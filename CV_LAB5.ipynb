```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
img = cv2.imread("/content/loki.jpg", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("error: image not found")
else:
    # Set up a grid for 8 bit planes + original image
    fig, axes = plt.subplots(3, 3, figsize=(10, 10))
    axes = axes.flatten()

    # Display Original Image
    axes[0].imshow(img, cmap="gray")
    axes[0].set_title("Original Image")
    axes[0].axis("off")

    # Extract bit-planes (Bit 0 to Bit 7)
    for i in range(8):
        # Bitwise AND with 2^i, then shift bits back to scale (or multiply by 255)
        bit_plane = (img >> i) & 1

        # Display each bit-plane (scaled to 0-255 for visualization)
        axes[i + 1].imshow(bit_plane * 255, cmap="gray")
        axes[i + 1].set_title(f"Bit Plane {i}")
        axes[i + 1].axis("off")

    plt.tight_layout()
    plt.show()
```
