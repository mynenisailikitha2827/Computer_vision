```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("/content/loki.jpg", cv2.IMREAD_GRAYSCALE)
if img is None:
    print("error: image not found")
else:
    fig, axes = plt.subplots(3, 3, figsize=(8,6))
    axes = axes.flatten()
    axes[0].imshow(img, cmap="gray")
    axes[0].set_title("Original Image")
    axes[0].axis("off")
    for i in range(8):
        bit_plane = (img >> i) & 1
        axes[i + 1].imshow(bit_plane * 255, cmap="gray")
        axes[i + 1].set_title(f"Bit Plane {i}")
        axes[i + 1].axis("off")
    plt.tight_layout()
    plt.show()
```
